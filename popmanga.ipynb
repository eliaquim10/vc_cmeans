```python
import os

def renomear_arquivos(arquivo_txt, diretorio):
    try:
        annotations_dir = os.path.join(diretorio, 'annotations/')
        images_dir = os.path.join(diretorio, 'images/')
        # print(annotations_dir)
        # print(images_dir)
        with open(arquivo_txt, 'r', encoding='utf-8') as file:
            for linha in file:
                base_file = linha.strip()
                nomes = base_file.split('(web) - ')
                # print(linha, len(linha))
                # print(nomes, len(nomes))
                sub_folder = os.path.dirname(nomes[0])
                new_file = nomes[1]

                image_dir = os.path.join(images_dir, base_file)
                json_dir = os.path.join(annotations_dir, base_file + '.json' )


                if len(nomes) == 2:
                    # l = len('[Unknown]')

                    # print('->', nomes[0])
                    # print('->', os.path.dirname(nomes[0]))
                    new_file = os.path.join(sub_folder, new_file)

                    new_image_dir = os.path.join(images_dir, new_file)
                    new_json_dir = os.path.join(annotations_dir, new_file + '.json' )


                    # print(image_dir, '\n', new_image_dir)
                    # print(json_dir, '\n', new_json_dir)
                    
                    # nome_antigo, nome_novo = nomes
                    # caminho_antigo = os.path.join(diretorio, nome_antigo)
                    # caminho_novo = os.path.join(diretorio, nome_novo)
                    
                    if os.path.exists(image_dir):
                        os.rename(image_dir, new_image_dir)
                        os.rename(json_dir, new_json_dir)
                        print(f'Renomeado: {image_dir} -> {new_image_dir}')
                    elif not os.path.exists(new_image_dir):
                        print(f"falhou {image_dir}")
                    else:
                        print(f'Arquivo não encontrado: {image_dir}')
                else:
                    pass
                    # print(f'Formato inválido na linha: {linha.strip()}')
    except Exception as e:
        print(f'Erro: {e}')

# Exemplo de uso
arquivo_txt = 'C:/Users/Eliaquim/Documents/mestrado/aulas/visao_computacional/vc_cmeans/data/poptest/annotations/seen.txt'  # Nome do arquivo que contém os nomes
caminho_diretorio = 'C:/Users/Eliaquim/Documents/mestrado/aulas/visao_computacional/vc_cmeans/data/poptest/'  # Diretório onde estão os arquivos
renomear_arquivos(arquivo_txt, caminho_diretorio)
```
